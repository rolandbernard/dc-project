```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
df = pd.read_csv("../../data/bz_hazard_landslide.csv")
ProfileReport(df, title="Profiling Report")
```

```python
for row in df.itertuples():
    if row.objectid != int(row.objectid):
        print(row.objectid)
    if row.code != int(row.code):
        print(row.code)
    if row.gml_id != f"UrbanPlan-HazardZonePlan-Landslides.{int(row.objectid)}":
        print(row.gml_id, row.objectid)
    if not row.geom.startswith("POLYGON"):
        print(row.geom)
    if row.istat_code != int(row.istat_code):
        print(row.istat_code)
    if row.gem_id != int(row.gem_id):
        print(row.gem_id)
    if row.gem_id != row.istat_code % 100:
        print(row.gem_id)
```
