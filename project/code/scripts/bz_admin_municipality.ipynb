```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
df = pd.read_csv("../../data/bz_admin_municipality.csv")
ProfileReport(df, title="Profiling Report")
```

```python
for row in df.itertuples():
    if row.id != int(row.id):
        print(row.id)
    if row.gml_id != f"Municipalities.{int(row.id)}":
        print(row.gml_id, row.id)
    if not row.geom.startswith("POLYGON"):
        print(row.geom)
    if row.istat_code != int(row.istat_code):
        print(row.istat_code)
    if row.ogc_fid != row.id:
        print(row.ogc_fid, row.id)
    if row.code != int(row.code):
        print(row.code)
    if row.code != row.istat_code % 1000:
        print(row.code, row.istat_code)
```
